```python
# Clean column names
df.columns = df.columns.str.lower().str.strip().str.replace(" ", "_")
df.head()
```

```python
# Frequency of Purchases mapping
frequency_map = {
    "weekly": 4,
    "bi-weekly": 2,
    "fortnightly": 2,
    "monthly": 1,
    "every 3 months": 0.33,
    "quarterly": 0.33,
    "annually": 0.083
}

df["frequency_of_purchases_num"] = (
    df["frequency_of_purchases"]
        .str.lower()
        .map(frequency_map)
)
```

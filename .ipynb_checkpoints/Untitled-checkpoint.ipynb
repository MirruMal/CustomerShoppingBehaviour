```python
# Clean column names
df.columns = df.columns.str.lower().str.strip().str.replace(" ", "_")
df.head()
```

```python
# Frequency of Purchases mapping
frequency_map = {
    "weekly": 4,
    "bi-weekly": 2,
    "fortnightly": 2,
    "monthly": 1,
    "every 3 months": 0.33,
    "quarterly": 0.33,
    "annually": 0.083
}

df["frequency_of_purchases_num"] = (
    df["frequency_of_purchases"]
        .str.lower()
        .map(frequency_map)
)
```

```python
df.info()
```

```python
df.head()
```

```python
df["frequency_of_purchases_num"].isna().sum()
```

```python
# Segmentation
segmentation_df = df[[
    "purchase_amount_(usd)",
    "previous_purchases",
    "frequency_of_purchases_num",
    "subscription_status",
    "season"
]].copy()
```

```python
segmentation_df = pd.get_dummies(
    segmentation_df,
    columns=["subscription_status", "season"],
    drop_first=True
)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaled_data = scaler.fit_transform(segmentation_df)
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

inertia = []

for k in range(1, 10):
    km = KMeans(n_clusters=k, random_state=42)
    km.fit(scaled_data)
    inertia.append(km.inertia_)

plt.plot(range(1, 10), inertia, marker="o")
plt.xlabel("Number of Clusters (k)")
plt.ylabel("Inertia")
plt.title("Elbow Method for Optimal k")
plt.show()
```

```python
kmeans = KMeans(n_clusters=3, random_state=42)
df["cluster"] = kmeans.fit_predict(scaled_data)
```

```python
df.groupby("cluster")[[
    "purchase_amount_(usd)",
    "previous_purchases",
    "frequency_of_purchases_num"
]].mean()
```

```python
df.groupby("cluster")["subscription_status"].value_counts(normalize=True)
```

```python
df.groupby("cluster")["season"].value_counts(normalize=True)
```

## Spending insights
* The correlation matrix shows no strong linear relationship between purchase amount and the numerical variables (age, previous purchases, frequency, review rating).
* Season Strongly Influences Spending - customers spend the most in Fall and Winter, with Fall showing the highest average purchase amount <br>
Business Impact: Increase advertising and product launches during Fall/Winter; run Summer promotions to boost spending
* Product Category Drives Purchase Amount - Footwear and Clothing produce the highest spending, while Outerwear results in smaller transactions<br>
Business Impact: Prioritize high-value categories for recommendations, marketing placement, and inventory decisions
* Discounts and Promo Codes Do NOT Increase Spending - Discount users spend almost the same amount as non-discount users<br>
Business Impact: Use discounts to drive conversion, not higher order values. To increase order size, rely on bundles or premium product suggestions instead

## Loyal customer
A customer is loyal if
* They are subscribed or
* They have above-average previous purchases

```python
threshold = df["previous_purchases"].median()
```

```python
df["loyal"] = (
    (df["subscription_status"] == "yes") |
    (df["previous_purchases"] > threshold)
).astype(int)
```

```python
df["loyal"].value_counts()
```

```python
features = df[[
    "age",
    "purchase_amount_(usd)",
    "previous_purchases",
    "frequency_of_purchases_num",
    "season",
    "payment_method",
    "shipping_type",
    "discount_applied",
    "promo_code_used"
]]
```

```python
features = pd.get_dummies(features, drop_first=True)
```

```python
target = df["loyal"]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    features, target, test_size=0.2, random_state=42
)
```

```python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier(max_depth=4, random_state=42)
model.fit(X_train, y_train)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

y_pred = model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))

print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))
```

```python
import pandas as pd

importance = pd.DataFrame({
    "feature": features.columns,
    "importance": model.feature_importances_
}).sort_values(by="importance", ascending=False)

importance
```

## Insights
* Customers who are subscribed, or
* Customers with above-median previous purchases<br>

were labelled as loyal (1), while others were labelled not loyal (0). <br>

Since previous purchase behavior is part of the loyalty definition, it became the dominant indicator in the model.<br>

After training the Decision Tree, we obtained:
* Accuracy: 100%
* A perfect confusion matrix (no misclassifications)
* Feature importance showing that previous_purchases explains nearly all of the model’s decisions.
<br><br>This result is expected, because the model effectively recovered the rule used to define loyalty.<br><br>In other words, previous_purchases fully predicts loyalty under our chosen definition, and additional features such as season, age, shipping type, and discount usage did not contribute to the prediction.<br><br>Customer loyalty in this dataset is almost entirely determined by repeat purchase behavior.<br>
Customers with higher numbers of previous purchases are consistently identified as loyal.
<br><br>
This aligns with real-world business intuition:<br>
Those who have purchased frequently in the past are the most likely to return again
